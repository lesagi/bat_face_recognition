```python
import pandas as pd

# Load the data from the CSV file
file_path = '/Users/MAC/Documents/bat_face_rec/face_recognition/data/validation/predictions.csv'  # Replace with the actual path to your CSV file
df = pd.read_csv(file_path)
```

```python
# Step 1: Calculate the sum of y_hat for each group
grouped_sum = df.groupby(['input1_name', 'input1_frame', 'input2_name'])['y_hat'].sum().reset_index(name='total_y_hat')

# Step 2: Merge this sum back to the original DataFrame
df_merged = pd.merge(df, grouped_sum, on=['input1_name', 'input1_frame', 'input2_name'])

# Step 3: Calculate the percentage
df_merged['y_hat_percentage'] = (df_merged['y_hat'] / df_merged['total_y_hat'])

# Optionally, if you want to drop the total_y_hat column
df_merged.drop(columns=['total_y_hat'], inplace=True)

# Display the DataFrame with the percentage
print(df_merged)

```

```python
# Save the result to a new CSV file if needed
df_merged.to_csv('highest_y_hat_per_combination.csv', index=False)
```
